```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.api as sm
from statsmodels.tsa.stattools import adfuller
from statsmodels.tsa.seasonal import seasonal_decompose
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf

data = pd.read_csv("https://confrecordings.ams3.digitaloceanspaces.com/JDS/EXPL2/Chp11/superstore.csv")
df = data.copy()

print(data.head())
print(data.shape)
data.info()

df['Order Date'] = pd.to_datetime(df['Order Date'])

df['order_date_year'] = df['Order Date'].apply(lambda x: x.year)
df['order_date_month'] = df['Order Date'].apply(lambda x: x.month)
df['order_date_year_month'] = df['Order Date'].apply(lambda x: x.strftime('%Y%m'))
df['order_date'] = df['Order Date'].apply(lambda x: x.strftime('%Y%m%d'))
df['order_date_day'] = df['Order Date'].apply(lambda x: x.day)
df['order_date_dayofweek'] = df['Order Date'].apply(lambda x: x.dayofweek)
df['order_date_dayofweek_name'] = df['Order Date'].apply(lambda x: x.strftime('%a'))

print(df.head())

df.groupby(['State']).size().plot(kind='bar', figsize=(18, 8))
plt.show()

df.groupby(['Region']).size().plot(kind='bar', figsize=(18, 8))
plt.show()

sns.lineplot(x=df['order_date_year_month'].value_counts().sort_index().index, y=df['order_date_year_month'].value_counts().sort_index().values)
plt.xticks(rotation=90)
plt.show()

sns.countplot(x=df['order_date_month'])
plt.show()

df.groupby(['City']).size().sort_values(ascending=False).head(20).plot(kind='bar', figsize=(18, 8))
plt.show()

city_sales = df[['City', 'Sales']].groupby('City', as_index=False)['Sales'].sum().sort_values('Sales', ascending=False)
print(city_sales.head(10))

sales_data = df[['Order Date', 'Category', 'Sales']]
print(sales_data.head(10))

office_supplies = df.loc[df['Category'] == 'Office Supplies']
print(df['Order Date'].min())
print(df['Order Date'].max())

office_supplies = office_supplies[['Order Date', 'Sales']]
print(office_supplies.head())

office_supplies = office_supplies.sort_values('Order Date')
office_supplies = office_supplies.groupby('Order Date')['Sales'].sum().reset_index()
office_supplies = office_supplies.set_index('Order Date')
print(office_supplies.head())

monthly_office = office_supplies['Sales'].resample('MS').mean()
print(monthly_office.head())

monthly_office.plot(figsize=(16, 7))
plt.title('Monthly office supplies sales')
plt.show()

furniture = df.loc[df['Category'] == 'Furniture']
furniture = furniture[['Order Date', 'Sales']]
print(furniture.head())

furniture = furniture.sort_values('Order Date')
furniture = furniture.groupby('Order Date')['Sales'].sum().reset_index()
furniture = furniture.set_index('Order Date')
print(furniture.head())

monthly_furniture = furniture['Sales'].resample('MS').mean()
print(monthly_furniture.head())

monthly_furniture.plot(figsize=(16, 7))
plt.title('Monthly furniture sales')
plt.show()

tech_supplies = df.loc[df['Category'] == 'Technology']
tech_supplies = tech_supplies[['Order Date', 'Sales']]
print(tech_supplies.head())

tech_supplies = tech_supplies.sort_values('Order Date')
tech_supplies = tech_supplies.groupby('Order Date')['Sales'].sum().reset_index()
tech_supplies = tech_supplies.set_index('Order Date')
print(tech_supplies.head())

monthly_technology = tech_supplies['Sales'].resample('MS').mean()
print(monthly_technology.head())

monthly_technology.plot(figsize=(16, 7))
plt.title('Monthly technology sales')
plt.show()

furniture = pd.DataFrame({'Order Date': monthly_furniture.index, 'Sales': monthly_furniture.values})
office = pd.DataFrame({'Order Date': monthly_office.index, 'Sales': monthly_office.values})
techno = pd.DataFrame({'Order Date': monthly_technology.index, 'Sales': monthly_technology.values})

store = furniture.merge(office, how='inner', on='Order Date')
store = store.rename(columns={'Sales_x': 'furniture_sales', 'Sales_y': 'office_sales'})
store = store.merge(techno, how='inner', on='Order Date')
store = store.rename(columns={'Sales': 'techno_Sales'})
store = store.rename(columns={'techno_Sales': 'techno_sales'})

print(store.head())

plt.figure(figsize=(16, 7))
plt.plot(store['Order Date'], store['furniture_sales'], 'b-', label='Furniture')
plt.plot(store['Order Date'], store['office_sales'], 'r-', label='Office Supplies')
plt.plot(store['Order Date'], store['techno_sales'], 'g-', label='technology Supplies')
plt.xlabel('Date')
plt.ylabel('Sales')
plt.title('Monthly Sales Furniture vs. Monthly Office Supplies vs Monthly technology sales')
plt.legend()
plt.show()

test_result = adfuller(store['techno_sales'])
print(test_result)

store['Seasonal First Difference technology'] = store['techno_sales'] - store['techno_sales'].shift(12)
store['Seasonal First Difference technology'].plot(figsize=(16, 7))
plt.show()

series = store.techno_sales
result = seasonal_decompose(series, model='additive', period=12)
result.plot()
plt.show()

fig = plt.figure(figsize=(12, 8))
ax1 = fig.add_subplot(211)
fig = sm.graphics.tsa.plot_acf(store['Seasonal First Difference technology'].dropna(), ax=ax1)
ax2 = fig.add_subplot(212)
fig = sm.graphics.tsa.plot_pacf(store['Seasonal First Difference technology'].dropna(), ax=ax2)
plt.show()

SarimaxModel = sm.tsa.statespace.SARIMAX(store.techno_sales, order=(1, 1, 1), seasonal_order=(1, 1, 2, 12))
SarimaxModel1 = SarimaxModel.fit()
print(SarimaxModel1.summary())

forecast1 = SarimaxModel1.predict(start=0, end=(len(store.techno_sales)) + 5, typ='levels').rename('Forecast')
print("Next Six Month Forecast:", forecast1[-6:])

store.techno_sales.plot(figsize=(18, 5), legend=True, title='Time Series Sales Forecasts')
forecast1.plot(legend=True, figsize=(18, 5))
plt.show()

forecast1_in = forecast1.iloc[:len(store.techno_sales)]
MAPE1 = np.mean(np.abs(store.techno_sales - forecast1_in) / store.techno_sales) * 100
print(round(100 - MAPE1, 2))

test_result = adfuller(store['office_sales'])
print(test_result)

series = store.office_sales
result = seasonal_decompose(series, model='additive', period=12)
result.plot()
CurrentFig = plt.gcf()
CurrentFig.set_size_inches(11, 8)
plt.show()

fig = plt.figure(figsize=(12, 8))
ax1 = fig.add_subplot(211)
fig = sm.graphics.tsa.plot_acf(store['office_sales'], ax=ax1)
ax2 = fig.add_subplot(212)
fig = sm.graphics.tsa.plot_pacf(store['office_sales'], ax=ax2)
plt.show()

SarimaxModel2 = sm.tsa.statespace.SARIMAX(store.office_sales, order=(1, 1, 1), seasonal_order=(1, 1, 2, 12))
SalesModel2 = SarimaxModel2.fit()
print(SalesModel2.summary())

forecast2 = SalesModel2.predict(start=0, end=(len(store.office_sales)) + 5, typ='levels').rename('Forecast')
print("Next Six Month Forecast:", forecast2[-6:])

store.office_sales.plot(figsize=(18, 5), legend=True, title='Time Series Sales Forecasts')
forecast2.plot(legend=True, figsize=(18, 5))
plt.show()

forecast2_in = forecast2.iloc[:len(store.office_sales)]
MAPE2 = np.mean(np.abs(store.office_sales - forecast2_in) / store.office_sales) * 100
print(round(100 - MAPE2, 2))

test_result = adfuller(store['furniture_sales'])
print(test_result)

series = store.furniture_sales
results = seasonal_decompose(series, model='additive', period=12)
results.plot()
CurrentFig = plt.gcf()
CurrentFig.set_size_inches(11, 8)
plt.show()

fig = plt.figure(figsize=(12, 8))
ax1 = fig.add_subplot(211)
fig = sm.graphics.tsa.plot_acf(store['furniture_sales'].iloc[13:], ax=ax1)
ax2 = fig.add_subplot(212)
fig = sm.graphics.tsa.plot_pacf(store['furniture_sales'].iloc[13:], ax=ax2)
plt.show()

SarimaxModel = sm.tsa.statespace.SARIMAX(store.furniture_sales, order=(5, 1, 10), seasonal_order=(1, 0, 0, 12))
SalesModel = SarimaxModel.fit()
print(SalesModel.summary())

forecast = SalesModel.predict(start=0, end=(len(store.furniture_sales)) + 5, typ='levels').rename('Forecast')
print("Next Six Month Forecast:", forecast[-6:])

store.furniture_sales.plot(figsize=(18, 5), legend=True, title='Time Series Sales Forecasts')
forecast.plot(legend=True, figsize=(18, 5))
plt.show()

forecast_in = forecast.iloc[:len(store.furniture_sales)]
MAPE = np.mean(np.abs(store.furniture_sales - forecast_in) / store.furniture_sales) * 100
print('#### Accuracy of model:', round(100 - MAPE, 2), '####')
```
